Challenge 2 🤺
Crear una tarjeta personal. El usuario (mediante input) puede elegir:

Crear datos nuevos
Modificar datos
Eliminar datos

```python
def validacion_rango(respuesta):
    while respuesta < 1 or respuesta > 4:
            respuesta = int(input("Recuerda debes seleccionar del 1 al 4 (1-Agregar dato, 2-Modificar dato, 3-Eliminar dato, 4-Salir). Responde nuevamente: ")) 
    return respuesta

def validacion_existencia(clave, diccionarios):
    while clave not in diccionarios:
        clave=input("La clave no existe. Igrese de nuevo:")
    return clave
    
diccionario={}

while True:
    print("Seleccione de 1 al 4 (1-Agregar dato, 2-Modificar dato, 3-Eliminar dato, 4-Salir)")
    seleccion= validacion_rango(int(input("Ingrese su eleccion:")))
    if seleccion==1:
        clave = input("¿Qué clave deseas agregar? (ej. 'nombre', 'edad'):").lower()
        valor = input("Ingrese el valor para el diccionario:").lower()
        diccionario[clave]= valor
        print(diccionario)
    elif seleccion==2:
        if len(diccionario)!=0:
            clave = validacion_existencia(input("¿Qué clave deseas modificar? (ej. 'nombre', 'edad'):").lower(), diccionario.keys())
            valor = input("Ingrese el nuevo valor para el diccionario:").lower()
            diccionario[clave]= valor
            print(diccionario)
        else:
            print("No cargo ningun dato debe de cargar alguno para poder modificarlo.")
    elif seleccion==3:
        if len(diccionario)!=0:
            clave =  validacion_existencia(input("¿Qué clave deseas eliminar? (ej. 'nombre', 'edad'):").lower(), diccionario.keys())
            del diccionario[clave]
            print(diccionario)
        else:
            print("No cargo ningun dato debe de cargar alguno para poder eliminarlo.")
    elif seleccion==4:
        print("Saliendo del programa")
        if len(diccionario)!=0:
            print(diccionario)
        else:
            print("No cargo ningun dato")
        break
```

Challenge 3 🤺
Crear una agenda de contactos. El usuario puede elegir:

Mostrar los contactos
Crear contacto
Modificar datos del contacto

```python
def validacion_rango(dato):
    while dato < 1 or dato > 4:
        dato = int(input("Recuerda debes seleccionar del 1 al 4 (1-Mostrar contactos, 2-Crear contacto, 3-Modificar contacto, 4-Salir). Responde nuevamente: ")) 
    return dato

def validacion_existencia(clave, diccionario):
    while clave not in diccionario:
        clave = input("La clave no existe. Ingrese de nuevo: ").lower()
    return clave

def validacion_rango_2(dato):
    while dato < 1 or dato > 2:
        dato = int(input("Recuerda debes seleccionar del 1 al 2 (1-Modificar el nombre del contacto, 2- Modificar el número del contacto). Responde nuevamente: ")) 
    return dato

libreta_de_contactos = {
    "mama": '123456789', 
    'hermana': '987654321', 
    'abuela': '456789123', 
    'abuelo': '321654987', 
    'tio': '654321789', 
    'primo': '789654123'
}

while True:
    print("\nSeleccione de 1 al 4 (1-Mostrar contactos, 2-Crear contacto, 3-Modificar contacto, 4-Salir)")
    seleccion = validacion_rango(int(input("Ingrese su elección: ")))

    if seleccion == 1:
        print("\nContactos actuales:")
        print(libreta_de_contactos)

    elif seleccion == 2:
        nombre_contacto = input("¿Qué contacto deseas agregar? (ej. 'Hermana', 'Madre'): ").lower()
        numero = input("Ingrese el número para agendar: ")
        libreta_de_contactos[nombre_contacto] = numero
        print("\nContacto agregado con éxito.")
        print(libreta_de_contactos)
        
    elif seleccion == 3:
        nombre_contacto = validacion_existencia(input("¿Qué contacto deseas modificar? (ej. 'Hermana', 'Madre'): ").lower(), libreta_de_contactos)
        seleccion_2 = validacion_rango_2(int(input("Seleccione (1-Modificar el nombre del contacto, 2-Modificar el número del contacto): ")))

        if seleccion_2 == 1:
            nombre_contacto_modificado = input("Ingrese el nuevo nombre del contacto para la agenda: ").lower()
            numero = libreta_de_contactos[nombre_contacto]
            del libreta_de_contactos[nombre_contacto]
            libreta_de_contactos[nombre_contacto_modificado] = numero
            print("\nNombre del contacto modificado con éxito.")
            print(libreta_de_contactos)
            
        elif seleccion_2 == 2:
            nuevo_numero = input("Ingrese el nuevo número para la agenda: ")
            libreta_de_contactos[nombre_contacto] = nuevo_numero
            print("\nNúmero del contacto modificado con éxito.")
            print(libreta_de_contactos)
            
    elif seleccion == 4:
        print("\nSaliendo del programa...")
        print("Contactos finales:")
        print(libreta_de_contactos)
        break
```

Challenge 🤺
Crear dos métodos para nuestro animal y ejecutar, debe ser algo específico de ese animal.

Ejemplo: "Cuando hablo, digo X".
Ejemplo: "Cuando me muevo, hago X".

```python
class Gato():
    cola= 1
    patas= 4
    ojos= 2

    def __init__(self,raza,color):
        self.raza= raza
        self.color= color
        
    def maullido(self):
        return "miauu miauu miauuuu"
    
    def arañazo(self):
        return "slapp"
    
    def bufido(self):
        return "gfffff"

milo= Gato("siames", "gris")

print(milo.bufido())
print(milo.arañazo())
print(milo.maullido())

    
```

Challenge 🤺
Definir una clase madre que herede a sus clases hijas dos atributos de instancia. Mostrar ambos atributos en pantalla desde los objetos de las clases hijas.

Observación: Editar las clases hijas para que no se inicialicen.

```python
class humano:
    manos=2
    piernas=2
    ojos=2
    
    def __init__(self,inteligencia,vida,defensa,ataque):
        self.inteligencia=inteligencia
        self.vida=vida
        self.defensa=defensa
        self.ataque=ataque
        
    def atacar(self):
        return "Atacando al enemigo"
    
    def retirar(self):
        return "Retirando del combate"
    
class guerrero(humano):
    def __init__(self,inteligencia,vida,defensa,ataque,espada):
        super().__init__(inteligencia,vida,defensa,ataque)
        self.arma=espada
        self.ataque= ataque*espada
        

class mago(humano):
    def __init__(self,inteligencia,vida,defensa,ataque,poder):
        super().__init__(inteligencia,vida,defensa,ataque)
        self.poder= poder
        self.inteligencia= inteligencia*poder
        
class arquero(humano):
    def __init__(self,inteligencia,vida,defensa,ataque,arco):
        super().__init__(inteligencia,vida,defensa,ataque)
        self.arco= arco
        self.defensa= defensa*arco

Henry= humano(10,100,50,20)
Lucas= guerrero(20,80,30,40,7)
Diana= mago(60,60,20,10,5)
Julian= arquero(40,50,90,5,5)

print("Henry:")
print("Inteligencia:", Henry.inteligencia)
print("Vida:", Henry.vida)
print("Defensa:", Henry.defensa)    
print("Ataque:", Henry.ataque, "\n")
print("Lucas:") 
print("Inteligencia:", Lucas.inteligencia)
print("Vida:", Lucas.vida)
print("Defensa:", Lucas.defensa)
print("Ataque:", Lucas.ataque, "\n")
print("Diana:")
print("Inteligencia:", Diana.inteligencia)
print("Vida:", Diana.vida)
print("Defensa:", Diana.defensa)
print("Ataque:", Diana.ataque, "\n")
print("Julian:")
print("Inteligencia:", Julian.inteligencia)
print("Vida:", Julian.vida)
print("Defensa:", Julian.defensa)
print("Ataque:", Julian.ataque, "\n")

print(Diana.atacar())
print(Diana.retirar())

```
