Challenge 🤺 - 15m
Solicitar al usuario que ingrese una frase.

Imprimir un mensaje cada vez que se encuentre una letra "o".

```python
frase= input('Escriba una frase:')
contador=0

for letra in frase:
    if letra == 'o' or letra == 'O':
        contador+= 1
        print('Hay mas jugo')

print('Se encontro',contador,'veces la letra "o".')
    
```

```python
marcas_carteras = ['hermes', 'louis vuitton', 'gucci', 'prada', 'chanel', 'prune']
agregar=0
while agregar==0:
    print('La lista de marcas es:', marcas_carteras)
    pregunta= input('Quiere agregar otra marca (responda si o no): ')
    if pregunta=='si':
        agregar=0
        marca_agregada= input('Ingrese la marca: ')
        marcas_carteras.append(marca_agregada)
    elif pregunta=='no':
        agregar=1
        for marca in marcas_carteras:
            longitud=len(marca)
            if longitud == 6:
                print(f'La marca {marca} tiene 6 letras')
            elif longitud > 6:
                print(f'La marca {marca} tiene más de 6 letras')
        
    
```

Challenge 🤺 - 30m
Escribir un programa en el que se pregunte al usuario por una temperatura (en grados Celsius), imprima la temperatura escrita, y luego clasificar la temperatura ingresada en los siguientes niveles:

Entre -10 y 18 grados, que imprima "Fresquete".
Entre 19 y 30 grados, que imprima "Calor'i".
Entre 31 y 45 grados, que imprima "Hakuuuuuu".
Para todo lo demás, imprimir "Ñamano mba'e!".

```python
def niveles_temperatura(temperatura):
    if temperatura >= -10 and temperatura <= 18:
        print("Fresquete")
    elif  temperatura >= 19  and temperatura <= 30:
        print("Calor'i")
    elif temperatura >= 31  and temperatura <= 45:
        print("Hakuuuuuu")
    else:
        print("Ñamano mba'e!")

temperatura=int(input("Digite la temperatura en Celsius: "))

print("La temperatura en Celsius es: ", temperatura)

niveles_temperatura(temperatura)


```
