Challenge 1: Registro de Tareas con Diccionarios
Crea un programa que gestione las tareas de una persona. Debe usar un diccionario
para almacenar las tareas con la siguiente estructura:
● tarea: nombre de la tarea.
● estado: estado de la tarea (pendiente, en progreso, completada).
El programa debe permitir al usuario:
1. Agregar una tarea nueva.
2. Modificar el estado de una tarea.
3. Eliminar una tarea.
4. Mostrar todas las tareas y su estado

```python
def validacion_rango(dato):
    while dato < 1 or dato > 5:
        dato = int(input("Recuerda debes seleccionar del 1 al 2 (1-Modificar el nombre de la tarea, 2-Modificar el estado de la tarea): Responde nuevamente: ")) 
    return dato

def validacion_existencia(clave, diccionario):
    while clave not in diccionario:
        clave = input("Esa tarea no existe. Intenta de nuevo: ").lower()
    return clave

def validacion_estado(dato):
    estado_valido = ["completado", "en proceso", "pendiente"]
    while dato not in estado_valido:
        dato = input("Estado inválido. Usa 'completado', 'en proceso' o 'pendiente': ").lower()
    return dato

def validacion_rango_2(dato):
    while dato < 1 or dato > 2:
        dato = int(input("Recuerda debes seleccionar del 1 al 2 (1-Modificar el nombre de la tarea, 2-Modificar el estado de la tarea): Responde nuevamente: ")) 
    return dato

Registro_de_tareas = {}

while True:
    print("\nSelecciona una opción: 1. Agregar tarea 2. Modificar tarea 3. Eliminar tarea 4. Mostrar tareas 5. Salir")
    seleccion = validacion_rango(int(input("Tu elección: ")))

    if seleccion == 1:
        tarea = input("¿Qué tarea deseas agregar? (ej. 'barrer', 'estudiar'): ").lower()
        estado = validacion_estado(input("Estado ('completado', 'en proceso', 'pendiente'): ").lower())
        Registro_de_tareas[tarea] = estado
        print("Tarea agregada con éxito.")
        print("Tareas actuales:", Registro_de_tareas)

    elif seleccion == 2:
        if len(Registro_de_tareas)!=0:
            tarea = validacion_existencia(input("¿Qué tarea deseas modificar?: ").lower(), Registro_de_tareas)
            seleccion_2 = validacion_rango_2(int(input("Seleccione (1-Modificar el nombre de la tarea, 2-Modificar el estado de la tarea):")))
            
            if seleccion_2 == 1:
                nombre_tarea_modificado = input("Ingrese el nuevo nombre de la tarea: ").lower()
                estado_actual = Registro_de_tareas[tarea]
                del Registro_de_tareas[tarea]
                Registro_de_tareas[nombre_tarea_modificado] = estado_actual
                print("\nNombre de la tarea modificado con éxito.")
                print(Registro_de_tareas)
                
            elif seleccion_2 == 2:
                nuevo_estado = validacion_estado(input("Nuevo estado ('completado', 'en proceso', 'pendiente'): ").lower())
                Registro_de_tareas[tarea] = nuevo_estado
                print("Estado actualizado.")
                print("Tareas actuales:", Registro_de_tareas)
        else:
            print("No hay tareas registradas. Agrega una primero.")

    elif seleccion == 3:
        if len(Registro_de_tareas)!=0:
            clave = validacion_existencia(input("¿Qué tarea deseas eliminar?: ").lower(), Registro_de_tareas)
            del Registro_de_tareas[clave]
            print("Tarea eliminada.")
            print("Tareas restantes:", Registro_de_tareas)
        else:
            print("No hay tareas para eliminar.")

    elif seleccion == 4:
        if len(Registro_de_tareas)!=0:
            print("Lista de tareas y sus estados:")
            print(Registro_de_tareas)
        else:
            print("No hay tareas registradas.")

    elif seleccion == 5:
        print("Saliendo del programa. Hasta la próxima.")
        if len(Registro_de_tareas)!=0:
            print("Lista de tareas y sus estados:")
            print(Registro_de_tareas)
        else:
            print("No hay tareas registradas.")
        break
```

Challenge 2: Adivina el Número 
Crea un juego simple en el que el usuario tiene que adivinar un número secreto. El 
usuario puede hacer 3 intentos. Cada vez que falle, el programa debe decir si el número 
secreto es mayor o menor al ingresado . Si el usuario adivina, debe mostrar 
"¡Felicidades! Adivinaste". 

```python
def validacion_rango(dato):
    while dato < 1 or dato > 10:
        dato = int(input("Recuerda debes seleccionar del 1 al 10. Responde nuevamente: \n ")) 
    return dato

numero_secreto=5
intentos=0

print("¡Bienvenido al juego de adivinanza! \n")
print("Tienes 3 intentos para adivinar el número secreto entre 1 y 10. \n")

while intentos<3:
    numero=validacion_rango(int(input("Adivina el número secreto (entre 1 y 10): \n")))
    if numero==numero_secreto:
        print("\n ¡Felicidades! Adivinaste el número.")
        break
    else:
        if numero<numero_secreto:
            print("El número secreto es mayor. \n")
        else:
            print("El número secreto es menor.\n")
        intentos+=1
        print("Intenta de nuevo.\n")
if intentos==3:
    print("Lo siento, has agotado tus intentos. El número era:", numero_secreto)
```

Challenge 3: La Clase Animal  
Crea una clase llamada Animal que tenga los atributos: 
● Nombre 
● edad 
La clase debe tener un método llamado presentarse() que imprima algo como: 
"¡Hola! Soy [nombre] y tengo [edad] años. Luego, crea instancias de un perro y 
un gato y haz que ambos se presenten.

```python
class Animal:
    patas=4
    ojos=2
    def __init__(self,nombre,edad):
        self.nombre= nombre
        self.edad=edad
    
    def presentarse(self):
        return f"¡Hola! Soy {self.nombre} y tengo {self.edad} años"
    
perro= Animal("Max", 2)
print(perro.presentarse())
gato= Animal("Mango", 4)
print(gato.presentarse())
```

Challenge 4: Peluche  
Crea una clase Peluche con los atributos: 
● nombre 
● tamaño 
La clase debe tener un método abrazar() que imprima algo como: "¡Es hora de 
abrazar a [nombre], el peluche de tamaño [tamaño]!" 
Luego, crea un peluche y haz que sea abrazado. 
Conceptos usados: 
● Clases, Atributos, Métodos.

```python
class Peluche():
    def __init__(self,nombre,tamaño):
        self.nombre= nombre
        self.tamaño= tamaño
        
    def abrazar(self):
        return f"¡¡Es hora de abrazar a {self.nombre} , el peluche de tamaño {self.tamaño}"
    
Osito= Peluche("teddy", "grande")
print(Osito.abrazar())    
```

Challenge 5: Multiplicar una Lista de Números  
Escribe una función llamada multiplicar_lista() que reciba una lista de números 
como parámetro y devuelva el resultado de multiplicar todos los elementos de la lista. 

```python
def multiplicar_lista(lista):
    resultado=1
    for numero in lista:
        resultado*=numero
    return resultado

Numeros= int(input("¿Cuántos números deseas multiplicar?: "))

for i in range(Numeros):
    numero= int(input("Ingresa un número: "))
    if i==0:
        Lista_numeros=[numero]
    else:
        Lista_numeros.append(numero)
        
print("La multiplicación de los números de la lista es:",multiplicar_lista(Lista_numeros))
```

Challenge 6: Acumulador de Números  
Escribe una función llamada acumulador() que reciba un número n como parámetro y 
devuelva la suma de todos los números desde 1 hasta n.

```python
def acumulador(numero):
    suma = 0
    for i in range(1, numero + 1):
        suma += i
    return suma

def validacion_nro_positivo(numero):
    while numero <= 0:
        numero = int(input("Recuerda debes ingresar un número positivo: "))
    return numero

numero_ingresado = validacion_nro_positivo(int(input("Ingresa un número para calcular la suma acumulada: ")))

print("El resultado de la suma acumulada es:", acumulador(numero_ingresado))
    
```
